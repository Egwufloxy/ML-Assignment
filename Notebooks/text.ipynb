```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Create some data
x = np.linspace(0, 10, 100)
y = np.sin(x)

# 2. Generate the plot
plt.figure(figsize=(8, 4))
plt.plot(x, y, label='Sine Wave', color='blue')

# 3. Add labels and a title
plt.title('My Simple Plot')
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.legend()
plt.grid(True)

# 4. Show the plot directly in the notebook
plt.show()
```
